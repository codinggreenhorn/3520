```python
import os
import pandas as pd
import statsmodels.api as sm

# Load playoff advanced + salary data
csv_path = 'data/playoff_adv_salary.csv' if os.path.exists('data/playoff_adv_salary.csv') else 'playoff_adv_salary.csv'
df = pd.read_csv(csv_path, encoding='latin1')

# Build MPG to match requested variable list
if 'MP' not in df.columns or 'G' not in df.columns:
    raise ValueError("Expected columns 'MP' and 'G' to build MPG.")

df['MPG'] = pd.to_numeric(df['MP'], errors='coerce') / pd.to_numeric(df['G'], errors='coerce')

x_cols = [
    'MPG', 'PER', 'TS%', '3PAr', 'FTr', 'ORB%', 'DRB%', 'TRB%', 'AST%', 'STL%',
    'BLK%', 'TOV%', 'USG%', 'OWS', 'DWS', 'WS', 'WS/48', 'OBPM', 'DBPM', 'BPM',
    'VORP', 'Pos_C', 'Pos_PF', 'Pos_PG', 'Pos_SF', 'Pos_SG', 'Awards'
]

# Target salary column (case-insensitive) and convert to millions
y_source = 'Salary' if 'Salary' in df.columns else ('salary' if 'salary' in df.columns else None)
if y_source is None:
    raise ValueError("Could not find salary column. Expected 'Salary' or 'salary'.")

df['salary_millions'] = pd.to_numeric(df[y_source], errors='coerce') / 1_000_000

# Keep required columns, coerce numeric, and drop missing rows
model_df = df[x_cols + ['salary_millions']].copy()
model_df = model_df.apply(pd.to_numeric, errors='coerce').dropna()

X = sm.add_constant(model_df[x_cols])
y = model_df['salary_millions']

model = sm.OLS(y, X).fit()
print(f'Loaded: {csv_path}')
print(model.summary())
```

```python
import os
import pandas as pd

csv_path = 'data/playoff_adv_salary.csv' if os.path.exists('data/playoff_adv_salary.csv') else 'playoff_adv_salary.csv'
df = pd.read_csv(csv_path, encoding='latin1')

if 'MP' not in df.columns or 'G' not in df.columns:
    raise ValueError("Expected columns 'MP' and 'G' to build MPG.")

mp_total = pd.to_numeric(df['MP'], errors='coerce')
games = pd.to_numeric(df['G'], errors='coerce')
df['MPG'] = mp_total / games

# Keep only players with MPG > 10
df = df[df['MPG'] > 10].copy()

print(f'Loaded: {csv_path}')
print(f'Rows with MPG > 10: {len(df)}')
df.head(20)
```

```python
import pandas as pd
from sklearn.ensemble import RandomForestRegressor

# Use filtered df from the previous cell (MPG > 10)
if 'df' not in globals():
    raise ValueError("Filtered dataframe 'df' not found. Run the previous cell first.")

feature_cols = [
    'MPG', 'PER', 'TS%', '3PAr', 'FTr', 'ORB%', 'DRB%', 'TRB%', 'AST%', 'STL%',
    'BLK%', 'TOV%', 'USG%', 'OWS', 'DWS', 'WS', 'WS/48', 'OBPM', 'DBPM', 'BPM',
    'VORP', 'Pos_C', 'Pos_PF', 'Pos_PG', 'Pos_SF', 'Pos_SG', 'Awards'
]

y_source = 'Salary' if 'Salary' in df.columns else ('salary' if 'salary' in df.columns else None)
if y_source is None:
    raise ValueError("Could not find salary column. Expected 'Salary' or 'salary'.")

rf_df = df[feature_cols + [y_source]].copy()
rf_df = rf_df.apply(pd.to_numeric, errors='coerce').dropna()

X = rf_df[feature_cols]
y = rf_df[y_source] / 1_000_000  # salary in millions

rf = RandomForestRegressor(n_estimators=500, random_state=42, n_jobs=-1)
rf.fit(X, y)

importance_df = pd.DataFrame({
    'feature': feature_cols,
    'importance': rf.feature_importances_
}).sort_values('importance', ascending=False)

print(f'Rows used for Random Forest: {len(rf_df)}')
print('Feature importances (highest to lowest):')
print(importance_df.to_string(index=False))
```
